```python
#syntax for function with arguments:
def evenOdd(x):
    if (x % 2 == 0):
        return "Even"
    else:
        return "Odd"
print (evenOdd(16))
print (evenOdd(7))
print (evenOdd(35))
```

```python
#Syntax for function with default argument:
def MyFunc(x, y = 50):
    print ("x:", x)
    print ("y:", y)
MyFunc(10)
```

```python
#Sample code for default argument:
def my_function(country = "Norway"):
    print ("I am from", country)
my_function("Sweden")
my_function("India")
my_function()
my_function("Brazil")
```

```python
#Keyword argument:
def student(fname, lname):
    print (fname, lname)
student (lname = "Man", fname = "Iron")
student (fname = "Iron", lname = "Man")
student("Iron", "Man")
```

```python
def nameAge(name, age):
    print ("Hi, I am", name)
    print ("My age is", age)
nameAge("Thor", 1500)
print ("\n")
nameAge(150, "Captain America")
```

```python
# 1
def calculate_sum(x, y, z):
    sum = x + y + z
    return sum

x = int(input("Enter the first number: "))
y = int(input("Enter the second number: "))
z = int(input("Enter the third number: "))

print(calculate_sum(x, y, z))
```
