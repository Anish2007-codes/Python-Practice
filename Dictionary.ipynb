```python
# 4
my_dict = {
    101 : "Amit",
    102 : "Rahul",
    103 : "Suraj"
}
print ("Keys:")
for key in my_dict:
    print(key)
print ("\nValues:")
for value in my_dict.values():
    print (value)
print ("\nKeys and Values:")
for key, value in my_dict.items():
    print (key,":",value)
```

```python
# 12
students = {
    "Amit" : 85,
    "Rahul" : 92,
    "Priya" : 78,
    "Sneha" : 95,
    "Neha" : 88
}
highest_student = max(students, key = students.get)
print ("Student with highest marks:", highest_student)
print ("Highest Marks:", students[highest_student])
```

```python
# 13
numbers = [10, 20, 10, 30, 20, 10, 40, 30]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1
print ("Frequency of elements:", frequency)
```

```python
# 14
my_dict = {
    "Amit" : 85,
    "Rahul" : 72,
    "Priya" : 95,
    "Sneha" : 80
}
sorted_dict1 = dict(sorted(my_dict.items(), key = lambda item : item[1]))
print ("Dictionary sorted by values:", sorted_dict1)

#For descending order:-
sorted_dict2 = dict(sorted(my_dict.items(), key = lambda item : item[1], reverse = True))
print ("Dictionary sorted in descending order:", sorted_dict2)
```
