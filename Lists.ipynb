```python
# 9
list1 = []
for i in range (20):
    x = int(input("Enter a number: "))
    list1.append(x)
print(f"This is the list : {list1}")
# a)
similar_elements = 0
element_index = []
for i in range(len(list1)):
    if list1.count(list1[i]) > 1:
        similar_elements += 1
        element_index.append(i)
print(f"The total no. of similar elements are : {similar_elements}")
print(f"The indices of the similar elements are : {element_index}")
# b) 
even_values = 0
odd_values = 0
for i in list1:
    if i % 2 == 0:
        even_values += 1
    else:
        odd_values += 1
print (f"The total no. of even values in the list are : {even_values}")
print (f"The total no. of odd values in the list are : {odd_values}")
#c
positive_value = 0
negative_value = 0
zero_value = 0
for i in list1:
    if i > 0:
        positive_value += 1
    elif i < 0:
        negative_value += 1
    else:
        zero_value += 1
print (f"The total no. of positive values in the list are : {positive_value}")
print (f"The total no. of negative values in the list are : {negative_value}")
print (f"The total no of values which are neither negative nor positive, that is, zeroes are : {zero_value}")
```

```python
# 18
list1 = [10, 20, 30, 40, 50, 60]
print (f"This is the first list : {list1}")
list2 = [10, 15, 20, 25, 30, 35, 40]
print (f"This is the second list : {list2}")
list3 = []
for i in list1:
    for j in list2:
        if i == j:
            list3.append(i)
print (f"The list of common elements is : {list3}")
```
