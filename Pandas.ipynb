```python
# Creating a Series from a dictionary - Example 2
import pandas
calories = {"day1" : 420, "day2" : 380, "day3" : 390}
print(pandas.Series(calories))
```

```python
# Datasets -
# Data sets in Pandas are usually multi-dimensional tables called DataFrames.
# Series is like a column whereas a DataFrame is the whole table.
import pandas
data = {
    "calories" : [420, 380, 390],
    "duration" : [50, 40, 45]
}
print (pandas.DataFrame(data))
```

```python
# We can use the loc attribute to return one or more specified rows - Example 1
import pandas
data = {
    "calories" : [420, 380, 390],
    "duration" : [50, 40, 45]
}
#load data into a DataFrame object:
df = pandas.DataFrame(data)
print (df.loc[0])
```

```python
# We can use the loc attribute to return one or more specified rows - Example 2
import pandas
data = {
    "calories" : [420, 380, 390],
    "duration" : [50, 40, 45]
}
#load data into a DataFrame object:
df = pandas.DataFrame(data)
print (df.loc[[0, 1]])
```

```python
# Named Indices
import pandas
data = {
    "calories" : [420, 380, 390],
    "duration" : [50, 40, 45]
}
df = pandas.DataFrame(data, index = ["day1", "day2", "day3"])
print (df)
```

```python
'''
Data Cleaning and preprocessing using Pandas:- 

Data cleaning means fixing and organizing messy data. 
Pandas offers a wide range of tools and functions to help us clean and preprocess our data effectively.

Data Cleaning often involves - 
1. Dropping irrelevant columns.
2. Renaming column names to meaningful names.
3. Making data values consisten.
4. Replacing or filling in missing values.
'''
```

```python
# Define a dictionary with sample data which includes some missing values

import pandas 
data = {
    'A' : [1, 2, 3, None, 5],
    'B' : [None, 2, 3, 4, 5],
    'C' : [1, 2, None, None, 5]
}
df = pandas.DataFrame(data)
print ("Original Data: \n", df)
print ("\n")

# use dropna() to remove rows with any missing values

df_cleaned = df.dropna()
print ("Cleaned Data: \n", df_cleaned)

# Here, we have used the dropna() method to remove rows with any missing values. 
# The resulting DataFrame df_cleaned will only contain rows without any missing values.
```

```python
# To fill the missing values in python, we use the fillna() function
import pandas 
data = {
    'A' : [1, 2, 3, None, 5],
    'B' : [None, 2, 3, 4, 5],
    'C' : [1, 2, None, None, 5]
}
df = pandas.DataFrame(data)
print ("Original Data: \n", df)
print ("\n")

# Filling NaN values with 0

df.fillna(0, inplace = True)
print ("Data after filling NaN with 0: \n", df)
```

```python
# fillna() function with mean() attribute
import pandas 
data = {
    'A' : [1, 2, 3, None, 5],
    'B' : [None, 2, 3, 4, 5],
    'C' : [1, 2, None, None, 5]
}
df = pandas.DataFrame(data)
print ("Original Data: \n", df)
print ("\n")

# Filling NaN values with the mean of each column
df.fillna(df.mean(), inplace = True)
print ("Data after filling NaN with mean: \n", df)
```

```python
# Handle Duplicate Values
import pandas as pd
# sample data
data = {
    'A' : [1, 2, 2, 3, 3, 4],
    'B' : [5, 6, 6, 7, 8, 8]
}
df = pd.DataFrame(data)
print ("Original DataFrame:\n", df.to_string(index = False))
# detect duplicates
print ("Duplicate Rows:\n", df[df.duplicated()].to_string(index = False))
# remove duplicates based on column 'A'
df.drop_duplicates(subset = ['A'], keep = 'first', inplace = True)
print ("DataFrame after removing duplicates based on column 'A': \n", df.to_string(index = False))

# 1. df[df.duplicated()] produces a boolean series to identify duplicate rows.
# 2. df.drop_duplicates(subset = ['A], keep = 'first', inplace = True) removes duplicates based on column A, retaining
# only the first occurrence of each duplicate directly in the original DataFrame.
```

```python
# Rename Column Names to Meaningful Names
# To rename column names to more meaningful names in Pandas, we can use the rename() function.
import pandas as pd
# sample data
data = {
    'A' : [25, 30, 35],
    'B' : ['John', 'Doe', 'Smith'],
    'C' : [50000, 60000, 70000]
}
df = pd.DataFrame(data)
# rename columns
df.rename(columns = {'A' : 'Age', 'B' : 'Name', 'C' : 'Salary'}, inplace = True)
print (df.to_string(index = False))

# Here, the columns of df are renamed from A, B and C to more meaningful names : Age, Name & Salary respectively.
```
