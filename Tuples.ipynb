```python
# 1
numbers = (10, 20, 30, 40, 50, 60, 70, 80, 90)
print ("Tuple:", numbers)
print ("4th element from the first:", numbers[3])
print ("4th element from the last:", numbers[-4])
```

```python
# 2
numbers = (10, 20, 30, 40, 50)
element = int(input("Enter element to search: "))
if element in numbers:
    print ("Element exists in the tuple!")
else:
    print ("Element does not exist in the tuple!")
```

```python
# 3
numbers_list = [10, 20, 30, 40, 50]
numbers_tuple = tuple(numbers_list)
print ("List:", numbers_list)
print ("Tuple:", numbers_tuple)
```

```python
# 4
numbers = (10, 20, 30, 40, 50)
element = int(input("Enter an element: "))
if element in numbers:
    print ("Index of element:", numbers.index(element))
else:
    print ("Element not found!")
```

```python
# 5
numbers = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
new_list = []
for tup in numbers:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)
print ("Original List:", numbers)
print ("Updated List:", new_list)
```

```python
# 6
numbers = (25, 10, 45, 5, 60, 30)
maximum = max(numbers)
minimum = min(numbers)
print (numbers)
print ("Maximum Number is:", maximum, "\nIndex is:", numbers.index(maximum))
print ("Minimum Number is:", minimum, "\nIndex is:", numbers.index(minimum))
```

```python
# 7
numbers = (10, 20, 30, 40, 50)
new_tuple = (numbers[-1],) + numbers[1:-1] + (numbers[0],)
print ("Original Tuple:", numbers)
print ("Tuple after swapping:", new_tuple)
```
